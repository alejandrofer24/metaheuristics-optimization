# 🐦 🐜 🐟  Particle Swarm Optimization (PSO)

## 1️⃣ What is PSO?

Particle Swarm Optimization (PSO) is a **population-based metaheuristic** inspired by the social behavior of birds flocking or fish schooling (**Bio-inspired**). It was first introduced by **James Kennedy and Russell Eberhart in 1995**.

An example can be a buch of **Birds** and we imagine that each bird is helping us to find the optimal solution, the best solution of the flock will be our best solution in the space.

This is a heuristic solution because we can never prove the real global optimal solution can be found and it is usually not. However, we often find that the solution found by PSO is quite close to the global optimal.

The idea is that a group of particles (candidate solutions) explore the solution space guided by:
- Their own best-known position (**personal best**, `pbest`)
- The best-known position found by the entire swarm (**global best**, `gbest`)

---

## 2️⃣ What is PSO used for?

PSO is used to solve **continuous optimization problems**, such as:
- Minimizing mathematical functions 
- Tuning hyperparameters of machine learning models
- Finding global optima in complex, non-linear or non-differentiable functions


---

## 3️⃣ Core Concepts and Equations

## Particle Attributes in PSO

Each particle has:

- **Position** $$ \vec{x}_i $$: A candidate solution.
- **Velocity** $$ \vec{v}_i $$: Direction and speed of movement.
- **Fitness value**: Objective function value at the current position.

Each particle stores:

- **Personal best** $$ \vec{p}_i $$: Best position found by the particle.
- **Global best** $$ \vec{g} $$: Best position found by the entire swarm.

---


### 🔁 Update Equations

**Velocity update:**
$$
\vec{v}_i(t+1) = w \cdot \vec{v}_i(t) + c_1 \cdot r_1 \cdot (\vec{p}_i - \vec{x}_i(t)) + c_2 \cdot r_2 \cdot (\vec{g} - \vec{x}_i(t))
$$

**Position update:**
$$
\vec{x}_i(t+1) = \vec{x}_i(t) + \vec{v}_i(t+1)
$$

Where:
-  w : inertia weight
-  c_1 : cognitive acceleration coefficient (self-confidence)
-  c_2 : social acceleration coefficient (swarm confidence)
-  r_1, r_2 - U(0, 1) : random values sampled from a uniform distribution

---

## 4️⃣ Algorithm example for minimizing some costfunction f(x)

for each particle i = 1, ..., S do

    Initialize the particle's position with a uniformly distributed random vector: xi ~ U(b_lo, b_up)
    Initialize the particle's best known position to its initial position: pi ← xi
    if f(pi) < f(g) then
        update the swarm's best known position: g ← pi
    Initialize the particle's velocity: vi ~ U(-|b_up-b_lo|, |b_up-b_lo|)

while a termination criterion is not met do:

    for each particle i = 1, ..., S do

        for each dimension d = 1, ..., n do

            Pick random numbers: rp, rg ~ U(0,1)
            Update the particle's velocity: vi,d ← w vi,d + φp rp (pi,d-xi,d) + φg rg (gd-xi,d)
        Update the particle's position: xi ← xi + vi
        

        if f(xi) < f(pi) then

            Update the particle's best known position: pi ← xi
            if f(pi) < f(g) then

                Update the swarm's best known position: g ← pi


Where:
-  w : inertia weight
-  φp : cognitive acceleration coefficient (self-confidence)
-  φg : social acceleration coefficient (swarm confidence)
-  r_p, r_g - U(0, 1) : random values sampled from a uniform distribution to the **best personal** and **global position**
- (pi,d-xi,d) : Vector pointing to the **best personal solution**
- (gd-xi,d) : Vector pointing to the **best global solution**

# First example
## Rastrigin Function

The **Rastrigin function** is a popular benchmark function in optimization, especially used to test the performance of metaheuristic algorithms like **Particle Swarm Optimization (PSO)**.

---

### Mathematical Formula

For a vector \( \mathbf{x} = (x_1, x_2, ..., x_n) \), the function is defined as:

$$
f(\mathbf{x}) = An + \sum_{i=1}^{n} \left[ x_i^2 - A \cos(2\pi x_i) \right]
$$

Where:

- \( A = 10 \)
- \( n \): number of dimensions
- \( x_i \in [-5.12, 5.12] \)

---

### Characteristics

- **Non-convex** and **multimodal** (many local minima)
- Global minimum at:

$$
\mathbf{x} = \mathbf{0} \quad \text{where} \quad f(\mathbf{0}) = 0
$$

-  Used to test the **exploration vs. exploitation** capabilities of optimization algorithms
-  Difficult for basic algorithms due to the large number of local optima

---

### Why Use It in PSO?

- Tests PSO’s ability to avoid getting trapped in **local minima**
- Good for visualizing **convergence behavior** in 2D or 3D
- Ideal to understand the balance between:
  - **Personal best (cognitive component)**
  - **Global best (social component)**

---

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the Rastrigin function
def rastrigin(x):
    A = 10
    return A * len(x) + np.sum(x**2 - A * np.cos(2 * np.pi * x))

# Class to represent a particle
class Particle:
    def __init__(self, dim, bounds, initial_position=None):
        if initial_position is not None:
            self.position = initial_position
        else:
            self.position = np.random.uniform(bounds[0], bounds[1], dim)

        self.velocity = np.random.uniform(-abs(bounds[1] - bounds[0]), abs(bounds[1] - bounds[0]), dim)
        self.best_position = self.position.copy()
        self.best_value = rastrigin(self.position)
# Particle Swarm Optimization with position tracking
def pso(n_particles=30, dim=2, bounds=(-5.12, 5.12), w=0.7, c1=1.5, c2=1.5, max_iter=100,max_velocity=None):

    
    side = int(np.sqrt(n_particles))
    x_vals = np.linspace(bounds[0], bounds[1], side)
    y_vals = np.linspace(bounds[0], bounds[1], side)
    xx, yy = np.meshgrid(x_vals, y_vals)
    positions = np.stack([xx.ravel(), yy.ravel()], axis=1)

    # Si n_particles no es cuadrado perfecto, corta o repite
    positions = positions[:n_particles]

    swarm = [Particle(dim, bounds, initial_position=pos) for pos in positions]  
    global_best_position = swarm[0].best_position.copy()
    global_best_value = swarm[0].best_value

    best_values = []
    positions_history = []  # Store positions for each iteration

    for t in range(max_iter):
        current_positions = []

        for particle in swarm:
            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)

            particle.velocity = (w * particle.velocity +
                                 c1 * r1 * (particle.best_position - particle.position) +
                                 c2 * r2 * (global_best_position - particle.position))
            
            # Aplicar límite si se especifica max_velocity
            if max_velocity is not None:
                particle.velocity = np.clip(particle.velocity, -max_velocity, max_velocity)



            particle.position += particle.velocity
            particle.position = np.clip(particle.position, bounds[0], bounds[1])

            value = rastrigin(particle.position)
            if value < particle.best_value:
                particle.best_value = value
                particle.best_position = particle.position.copy()

                if value < global_best_value:
                    global_best_value = value
                    global_best_position = particle.position.copy()

            current_positions.append(particle.position.copy())

        best_values.append(global_best_value)
        positions_history.append(np.array(current_positions))  # Save positions this iteration

    return global_best_position, global_best_value, best_values, positions_history
```

```python
# Create Rastrigin contour background
x = np.linspace(-5.12, 5.12, 400)
y = np.linspace(-5.12, 5.12, 400)
X, Y = np.meshgrid(x, y)
Z = 10 * 2 + (X**2 - 10 * np.cos(2 * np.pi * X)) + (Y**2 - 10 * np.cos(2 * np.pi * Y))

fig, ax = plt.subplots(figsize=(6, 6))
contour = ax.contourf(X, Y, Z, levels=50, cmap='viridis')
particles_plot, = ax.plot([], [], 'ro', markersize=3)
ax.set_title('Particle Swarm Optimization - Rastrigin Function')
ax.set_xlim(-5.12, 5.12)
ax.set_ylim(-5.12, 5.12)
```

```python
from matplotlib.animation import FuncAnimation

# Run PSO and get positions history
_, _, _, history = pso()

# Update function for each frame
def update(frame):
    positions = history[frame]
    particles_plot.set_data(positions[:, 0], positions[:, 1])
    ax.set_title(f'Iteration {frame + 1}')
    return particles_plot,

# Create animation
ani = FuncAnimation(fig, update, frames=len(history), interval=100, blit=True)

# Display animation in notebook
from IPython.display import HTML
HTML(ani.to_jshtml())
```

## Hyperparameters testing for example 1

### Key Hyperparameters and Their Roles

- **Number of Particles (`n_particles`)**  
  Controls the size of the swarm. More particles provide better exploration but increase computational cost.

- **Inertia Weight (`w`)**  
  Balances exploration and exploitation by scaling the previous velocity.  
  - High `w` promotes exploration (searching new areas).  
  - Low `w` promotes exploitation (fine-tuning around the best solutions).

- **Cognitive Coefficient (`c1`)**  
  Controls how much a particle is influenced by its own best-known position.

- **Social Coefficient (`c2`)**  
  Controls how much a particle is influenced by the swarm’s global best-known position.

- **Maximum Velocity (`v_max`)**  
  Limits the maximum velocity a particle can have to prevent it from moving too fast and skipping good solutions.

- **Maximum Iterations (`max_iter`)**  
  Number of iterations the algorithm runs, affecting convergence time and solution quality.


We will analyze how these affect convergence speed and final solution quality.


```python
def run_and_plot(particle_count=36, w=0.7, c1=1.5, c2=1.5, max_iter=100,max_velocity=None):
    _, _, best_values, positions_history = pso(
        n_particles=particle_count, dim=2,
        bounds=(-5.12,5.12),
        w=w, c1=c1, c2=c2,
        max_iter=max_iter,
        max_velocity=max_velocity
    )

    # Plot convergence
    plt.figure(figsize=(6, 4))
    plt.plot(best_values, label=f'Particles={particle_count}, w={w}, c1={c1}, c2={c2}')
    plt.xlabel('Iteration')
    plt.ylabel('Best Value Found')
    plt.title('Convergence Curve')
    plt.grid(True)
    plt.legend()
    plt.show()

    return best_values
```

```python
for n in [16, 36, 64]:
    run_and_plot(particle_count=n)
```

##  Ackley Function – Description and Properties

The **Ackley function** is a well-known benchmark function used to evaluate the performance of optimization algorithms, especially those that must handle **many local minima**.

###  Mathematical Definition (for dimension *n*):

$$
f(\mathbf{x}) = -a \cdot \exp\left(-b \cdot \sqrt{\frac{1}{n} \sum_{i=1}^{n} x_i^2} \right) - \exp\left( \frac{1}{n} \sum_{i=1}^{n} \cos(c \cdot x_i) \right) + a + \exp(1)
$$

Typical parameter values:
- \( a = 20 \)
- \( b = 0.2 \)
- \( c = 2\pi \)

---

### 🌍 Global Characteristics

| Property              | Description                             |
|-----------------------|-----------------------------------------|
| **Dimension**         | Any \( n \), commonly 2 or 10           |
| **Search space**      | \( x_i \in [-5, 5] \)                   |
| **Global minimum**    | \( \mathbf{x} = \mathbf{0} \), \( f = 0 \) |
| **Local minima**      | Many, distributed throughout the space  |
| **Landscape**         | Flat central basin with rippling edges  |
| **Difficulty**        | High, due to deceptive landscape        |

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the Rastrigin function
def ackley(x, a=20, b=0.2, c=2 * np.pi):
    x = np.array(x)
    n = len(x)
    sum_sq_term = -a * np.exp(-b * np.sqrt(np.sum(x**2) / n))
    cos_term = -np.exp(np.sum(np.cos(c * x)) / n)
    return sum_sq_term + cos_term + a + np.exp(1)

# Class to represent a particle
class Particle:
    def __init__(self, dim, bounds, initial_position=None):
        if initial_position is not None:
            self.position = initial_position
        else:
            self.position = np.random.uniform(bounds[0], bounds[1], dim)

        self.velocity = np.random.uniform(-abs(bounds[1] - bounds[0]), abs(bounds[1] - bounds[0]), dim)
        self.best_position = self.position.copy()
        self.best_value = rastrigin(self.position)
# Particle Swarm Optimization with position tracking
def pso2(n_particles=30, dim=2, bounds=(-5.12, 5.12), w=0.7, c1=1.5, c2=1.5, max_iter=100,max_velocity=None):

    
    side = int(np.sqrt(n_particles))
    x_vals = np.linspace(bounds[0], bounds[1], side)
    y_vals = np.linspace(bounds[0], bounds[1], side)
    xx, yy = np.meshgrid(x_vals, y_vals)
    positions = np.stack([xx.ravel(), yy.ravel()], axis=1)

    # Si n_particles no es cuadrado perfecto, corta o repite
    positions = positions[:n_particles]

    swarm = [Particle(dim, bounds, initial_position=pos) for pos in positions]  
    global_best_position = swarm[0].best_position.copy()
    global_best_value = swarm[0].best_value

    best_values = []
    positions_history = []  # Store positions for each iteration

    for t in range(max_iter):
        current_positions = []

        for particle in swarm:
            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)

            particle.velocity = (w * particle.velocity +
                                 c1 * r1 * (particle.best_position - particle.position) +
                                 c2 * r2 * (global_best_position - particle.position))
            
            # Aplicar límite si se especifica max_velocity
            if max_velocity is not None:
                particle.velocity = np.clip(particle.velocity, -max_velocity, max_velocity) 



            particle.position += particle.velocity
            particle.position = np.clip(particle.position, bounds[0], bounds[1])

            value = ackley(particle.position)
            if value < particle.best_value:
                particle.best_value = value
                particle.best_position = particle.position.copy()

                if value < global_best_value:
                    global_best_value = value
                    global_best_position = particle.position.copy()

            current_positions.append(particle.position.copy())

        best_values.append(global_best_value)
        positions_history.append(np.array(current_positions))  # Save positions this iteration

    return global_best_position, global_best_value, best_values, positions_history
```

```python
# Create Rastrigin contour background
x = np.linspace(-5.12, 5.12, 400)
y = np.linspace(-5.12, 5.12, 400)
X, Y = np.meshgrid(x, y)
Z = -20 * np.exp(-0.2 * np.sqrt(0.5 * (X**2 + Y**2))) \
    - np.exp(0.5 * (np.cos(2 * np.pi * X) + np.cos(2 * np.pi * Y))) \
    + 20 + np.e

fig, ax = plt.subplots(figsize=(6, 6))
contour = ax.contourf(X, Y, Z, levels=50, cmap='viridis')
particles_plot, = ax.plot([], [], 'ro', markersize=3)
ax.set_title('Particle Swarm Optimization - Rastrigin Function')
ax.set_xlim(-5.12, 5.12)
ax.set_ylim(-5.12, 5.12)
```

```python
from matplotlib.animation import FuncAnimation

# Run PSO and get positions history
_, _, _, history = pso2()

# Update function for each frame
def update(frame):
    positions = history[frame]
    particles_plot.set_data(positions[:, 0], positions[:, 1])
    ax.set_title(f'Iteration {frame + 1}')
    return particles_plot,

# Create animation
ani = FuncAnimation(fig, update, frames=len(history), interval=100, blit=True)

# Display animation in notebook
from IPython.display import HTML
HTML(ani.to_jshtml())
```

```python
def pso2(n_particles=36, dim=2, bounds=(-5.12, 5.12), w=0.7, c1=1.5, c2=1.5, max_iter=100,max_velocity=0.5):

    
    side = int(np.sqrt(n_particles))
    x_vals = np.linspace(bounds[0], bounds[1], side)
    y_vals = np.linspace(bounds[0], bounds[1], side)
    xx, yy = np.meshgrid(x_vals, y_vals)
    positions = np.stack([xx.ravel(), yy.ravel()], axis=1)

    # Si n_particles no es cuadrado perfecto, corta o repite
    positions = positions[:n_particles]

    swarm = [Particle(dim, bounds, initial_position=pos) for pos in positions]  
    global_best_position = swarm[0].best_position.copy()
    global_best_value = swarm[0].best_value

    best_values = []
    positions_history = []  # Store positions for each iteration

    for t in range(max_iter):
        current_positions = []

        for particle in swarm:
            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)

            particle.velocity = (w * particle.velocity +
                                 c1 * r1 * (particle.best_position - particle.position) +
                                 c2 * r2 * (global_best_position - particle.position))
            
            # Aplicar límite si se especifica max_velocity
            if max_velocity is not None:
                particle.velocity = np.clip(particle.velocity, -max_velocity, max_velocity)



            particle.position += particle.velocity
            particle.position = np.clip(particle.position, bounds[0], bounds[1])

            value = ackley(particle.position)
            if value < particle.best_value:
                particle.best_value = value
                particle.best_position = particle.position.copy()

                if value < global_best_value:
                    global_best_value = value
                    global_best_position = particle.position.copy()

            current_positions.append(particle.position.copy())

        best_values.append(global_best_value)
        positions_history.append(np.array(current_positions))  # Save positions this iteration

    return global_best_position, global_best_value, best_values, positions_history
```

##  Benchmark Functions Used

###  Rastrigin Function

- **Formula**:  
  $$
  f(x) = An + \sum_{i=1}^n [x_i^2 - A \cos(2 \pi x_i)], \quad A = 10
  $$
- **Characteristics**:
  - Highly **multimodal** (many local minima).
  - Global minimum at **x = 0**, where **f(x) = 0**.
  - Challenging for optimization due to deceptive local minima.

###  Ackley Function

- **Formula**:  
  $$
  f(x) = -a \cdot \exp\left(-b \cdot \sqrt{\frac{1}{n} \sum x_i^2}\right) - \exp\left(\frac{1}{n} \sum \cos(c x_i)\right) + a + e
  $$
  Typically:  
  $a = 20$, $b = 0.2$, $c = 2\pi$
- **Characteristics**:
  - Multimodal, but **smoother** landscape.
  - Global minimum also at **x = 0**, where **f(x) = 0**.
  - Easier to optimize compared to Rastrigin.
